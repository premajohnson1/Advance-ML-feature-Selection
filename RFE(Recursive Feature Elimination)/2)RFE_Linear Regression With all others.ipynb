```python

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import RFE

from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import r2_score, mean_squared_error


# 1. Load dataset
dataset = pd.read_csv("Preplacementdata.csv")
dataset=dataset.drop('sl_no',axis=1)

# 2. Convert categorical columns into numerical columns
df2 = pd.get_dummies(dataset, drop_first=True, dtype=int)


# 3. Separate independent and dependent variables
indep_X = df2.drop('salary', axis=1)
dep_Y = df2['salary']


# 4. Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    indep_X, dep_Y,
    test_size=0.25,
    random_state=0
)


# 5. Feature Scaling
sc = StandardScaler()

X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)


# 6. Create Linear Regression model for RFE
lin = LinearRegression()


# 7. Apply RFE
rfe = RFE(
    estimator=lin,
    n_features_to_select=3
)


# 8. Fit RFE using training data
rfe.fit(X_train, y_train)


# 9. Display selected features
selected_features = rfe.support_

print("Selected Features:")
print(indep_X.columns[selected_features])


# 10. Display feature rankings
feature_df = pd.DataFrame({
    'Feature': indep_X.columns,
    'Selected': rfe.support_,
    'Ranking': rfe.ranking_
})

print("\nFeature Ranking:")
print(feature_df)


# 11. Transform training and testing data
X_train_selected = rfe.transform(X_train)
X_test_selected = rfe.transform(X_test)


# 12. Create multiple regression algorithms
regression_models = {
    'Linear Regression': LinearRegression(),

    'SVM Linear': SVR(kernel='linear'),

    'SVM Non-Linear': SVR(kernel='rbf'),

    'Decision Tree': DecisionTreeRegressor(
        random_state=0
    ),

    'Random Forest': RandomForestRegressor(
        n_estimators=10,
        random_state=0
    )
}


# 13. Train and evaluate all algorithms
result = []

for name, model in regression_models.items():

    # Train model
    model.fit(X_train_selected, y_train)

    # Predict test data
    y_pred = model.predict(X_test_selected)

    # Calculate evaluation metrics
    r2 = r2_score(y_test, y_pred)

    mse = mean_squared_error(y_test, y_pred)

    result.append({
        'Algorithm': name,
        'R2 Score': r2,
        'MSE': mse
    })


# 14. Display comparison results
result_df = pd.DataFrame(result)

print("\nRegression Algorithm Comparison:")
print(result_df)


# 15. Identify the best model based on R2 Score
best_model = result_df.loc[
    result_df['R2 Score'].idxmax()
]

print("\nBest Algorithm:")
print(best_model)
```

```python
- R²: Higher is better.
- MSE: Lower is better.
The best model in this illustrative example is Random Forest.

Remember these two important points:
1. RFE is performed only once using Linear Regression.
2. All five algorithms are trained separately using the same three selected features.
This allows you to compare different regression algorithms without repeating RFE for every algorithm.
```
